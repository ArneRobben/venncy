# Pillow renderer

Lightweight raster Venn diagrams with **Pillow** — no matplotlib required.

```bash
pip install venncy[pillow]
```

````python
from venncy.renderers.pillow import plot_venn2

# Just provide the three areas — the renderer does the rest
img = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
img
````

## No overlap

````python
img = plot_venn2(area_a=10, area_b=3, area_ab=0)
img
````

## Full containment

````python
img = plot_venn2(area_a=10, area_b=3, area_ab=3)
img
````

## Customisation

You can pass custom colours, opacity (0–255), dimensions, and background colour.

````python
img = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("green", "red"),
    alpha=60,
    width=700,
    height=500,
)
img
````

## Save to file

Export the image as PNG.

````python
img = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("A", "B"))
img.save("venn_example.png")
print("Saved venn_example.png")
````
