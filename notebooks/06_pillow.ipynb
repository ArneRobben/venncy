# Pillow Renderer

This notebook demonstrates how to create Venn diagrams using the **Pillow** renderer.

**What you'll learn:**
1. Drawing 2-circle Venn diagrams (`plot_venn2`)
2. Drawing 3-circle Venn diagrams (`plot_venn3`)
3. Customising colours, labels, and layout

```bash
pip install venncy[pillow]
```

````python
from venncy.renderers.pillow import plot_venn2

# Just provide the three areas — the renderer does the rest
img = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
img
````

### No overlap

````python
img = plot_venn2(area_a=10, area_b=3, area_ab=0)
img
````

### Full containment

````python
img = plot_venn2(area_a=10, area_b=3, area_ab=3)
img
````

### Customisation

You can pass custom colours, opacity (0–255), dimensions, background colour, and any extra `**kwargs` forwarded to `ImageDraw.ellipse()`. Use `text_kwargs` to customise the label.

````python
img = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("green", "red"),
    alpha=60,
    width=700,
    height=500,
    text_kwargs={"fill": "darkblue"},
)
img
````

## 3-Circle Venn Diagram

### Basic usage

For three circles, provide six area values — the three individual areas (`area_a`, `area_b`, `area_c`) and the three pairwise overlaps (`area_ab`, `area_ac`, `area_bc`).

````python
from venncy.renderers.pillow import plot_venn3

img = plot_venn3(area_a=10, area_b=8, area_c=6, area_ab=4, area_ac=3, area_bc=2)
img
````

### Customisation

Same options as `plot_venn2` — pass custom `colors`, `alpha`, or any `**kwargs` for `ImageDraw.ellipse()`.

````python
img = plot_venn3(
    area_a=10, area_b=8, area_c=6,
    area_ab=4, area_ac=3, area_bc=2,
    labels=("A", "B", "C"),
    colors=("orange", "purple", "cyan"),
    alpha=80,
    text_kwargs={"fill": "darkblue"},
)
img
````

## Save to file

Export the image as PNG.

````python
img = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("A", "B"))
img.save("venn_example.png")
print("Saved venn_example.png")
````
