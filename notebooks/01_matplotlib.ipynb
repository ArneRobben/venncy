## Customisation

You can pass an existing `Axes`, custom colours, or adjust alpha.

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4), dpi=120)

from venncy.renderers.matplotlib import plot_venn2

plot_venn2(10, 5, 2, ax=axes[0], labels=("A", "B"))
axes[0].set_title("Partial overlap")

plot_venn2(10, 5, 0, ax=axes[1], colors=("tab:green", "tab:red"), labels=("X", "Y"))
axes[1].set_title("No overlap")

plot_venn2(10, 3, 3, ax=axes[2], alpha=0.5, labels=("Outer", "Inner"))
axes[2].set_title("Full containment")

plt.tight_layout()
```
