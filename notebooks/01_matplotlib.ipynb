# Matplotlib Renderer

This notebook demonstrates how to create Venn diagrams using the **matplotlib** renderer.

**What you'll learn:**
1. Drawing 2-circle Venn diagrams (`plot_venn2`)
2. Drawing 3-circle Venn diagrams (`plot_venn3`)
3. Customising colours, opacity, labels, and layout

## 2-Circle Venn Diagram

### Basic usage

Provide the three area values — `area_a`, `area_b`, and their overlap `area_ab` — and the renderer calculates circle sizes and positioning automatically.

```python
from venncy.renderers.matplotlib import plot_venn2

# Just provide the three areas — the renderer does the rest
fig, ax = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
```

### No overlap

When `area_ab=0` the circles are drawn side by side with no intersection.

```python
fig, ax = plot_venn2(area_a=10, area_b=3, area_ab=0)
```

### Full containment

When `area_ab` equals the smaller area, the small circle is fully contained within the larger one.

```python
fig, ax = plot_venn2(area_a=10, area_b=3, area_ab=3)
```

### Customisation

You can pass an existing matplotlib `Axes`, custom colours, and adjust opacity (`alpha`). This makes it easy to compose multiple diagrams in a single figure.

```python
import matplotlib.pyplot as plt
from venncy.renderers.matplotlib import plot_venn2

fig, axes = plt.subplots(1, 3, figsize=(15, 4), dpi=120)

plot_venn2(
    area_a=10, 
    area_b=5, 
    area_ab=2, 
    ax=axes[0],
    labels=("A", "B"),
    colors=("tab:blue", "tab:orange"),
    alpha=0.5, 
    text_kwargs={"fontsize": 12}
    )
axes[0].set_title("Partial overlap")

plot_venn2(
    area_a=10, 
    area_b=5, 
    area_ab=0, 
    ax=axes[1], 
    labels=("X", "Y"),
    colors=("tab:green", "tab:red"),
    alpha=0.5,
    text_kwargs={"fontsize": 12}
)
axes[1].set_title("No overlap")

plot_venn2(
    area_a=10, 
    area_b=3, 
    area_ab=3, 
    ax=axes[2], 
    labels=("Outer", "Inner"),
    colors=("tab:purple", "tab:cyan"),
    alpha=0.5,
    text_kwargs={"fontsize": 12, "ha": "right", "va": "bottom"}
)
axes[2].set_title("Full containment")

plt.tight_layout()
```

## 3-Circle Venn Diagram

### Basic usage

For three circles, provide six area values — the three individual areas (`area_a`, `area_b`, `area_c`) and the three pairwise overlaps (`area_ab`, `area_ac`, `area_bc`).

```python
from venncy.renderers.matplotlib import plot_venn3

fig, ax = plot_venn3(area_a=10, area_b=8, area_c=6, area_ab=4, area_ac=3, area_bc=2)
```

### Customisation

Same options as `plot_venn2` — pass an existing `Axes`, custom `colors`, or adjust `alpha`.

```python
import matplotlib.pyplot as plt
from venncy.renderers.matplotlib import plot_venn3

fig, axes = plt.subplots(1, 2, figsize=(15, 4), dpi=120)

plot_venn3(
    area_a=10, area_b=8, area_c=6,
    area_ab=4, area_ac=3, area_bc=2,
    ax=axes[0],
    labels=("A", "B", "C"),
    colors=("tab:orange", "tab:purple", "tab:cyan"),
    alpha=0.7,
    edgecolor="black",
    text_kwargs={"fontsize": 12},
)
axes[0].set_title("Partial overlap")

plot_venn3(
    area_a=10, area_b=8, area_c=6,
    area_ab=0, area_ac=0, area_bc=0,
    ax=axes[1],
    labels=("X", "Y", "Z"),
    colors=("tab:green", "tab:red", "tab:blue"),
    alpha=0.7,
    edgecolor="black",
    text_kwargs={"fontsize": 12, "font": "monospace"},
)
axes[1].set_title("No overlap")

plt.tight_layout()
```
