# Plotly Renderer

This notebook demonstrates how to create Venn diagrams using the **Plotly** renderer.

**What you'll learn:**
1. Drawing 2-circle Venn diagrams (`plot_venn2`)
2. Drawing 3-circle Venn diagrams (`plot_venn3`)
3. Customising colours, labels, and layout

```bash
pip install venncy[plotly]
```

````python
from venncy.renderers.plotly import plot_venn2

fig = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
fig.show()
````

### No overlap

````python
fig = plot_venn2(area_a=10, area_b=3, area_ab=0)
fig.show()
````

### Full containment

````python
fig = plot_venn2(area_a=10, area_b=3, area_ab=3)
fig.show()
````

### Customisation

You can pass custom colours, line colours, figure dimensions, and any extra `**kwargs` forwarded to `fig.add_shape()`. Use `text_kwargs` to customise the label font.

````python
fig = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("rgba(50,200,50,0.3)", "rgba(200,50,50,0.3)"),
    line_colors=("green", "red"),
    width=700,
    height=500,
    line_dash="dash",
    text_kwargs={"size": 20, "color": "darkblue"},
)
fig.show()
````

## 3-Circle Venn Diagram

### Basic usage

For three circles, provide six area values — the three individual areas (`area_a`, `area_b`, `area_c`) and the three pairwise overlaps (`area_ab`, `area_ac`, `area_bc`).

````python
from venncy.renderers.plotly import plot_venn3

fig = plot_venn3(area_a=10, area_b=8, area_c=6, area_ab=4, area_ac=3, area_bc=2)
fig.show()
````

### Customisation

Same options as `plot_venn2` — pass custom `colors`, `line_colors`, or any `**kwargs` for `fig.add_shape()`.

````python
fig = plot_venn3(
    area_a=10, area_b=8, area_c=6,
    area_ab=4, area_ac=3, area_bc=2,
    labels=("A", "B", "C"),
    colors=("rgba(255,165,0,0.35)", "rgba(128,0,128,0.35)", "rgba(0,255,255,0.35)"),
    line_colors=("orange", "purple", "cyan"),
    line_dash="dot",
    text_kwargs={"size": 18},
)
fig.show()
````
