# Plotly renderer

Interactive Venn diagrams with **Plotly**.

```bash
pip install venncy[plotly]
```

````python
from venncy.renderers.plotly import plot_venn2

fig = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
fig.show()
````

## No overlap

````python
fig = plot_venn2(area_a=10, area_b=3, area_ab=0)
fig.show()
````

## Full containment

````python
fig = plot_venn2(area_a=10, area_b=3, area_ab=3)
fig.show()
````

## Customisation

You can pass custom colours, line colours, and figure dimensions.

````python
fig = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("rgba(50,200,50,0.3)", "rgba(200,50,50,0.3)"),
    line_colors=("green", "red"),
    width=700,
    height=500,
)
fig.show()
````
