# pyvis Renderer

This notebook demonstrates how to create Venn diagrams using the **pyvis** renderer.

**What you'll learn:**
1. Drawing 2-circle Venn diagrams (`plot_venn2`)
2. Drawing 3-circle Venn diagrams (`plot_venn3`)
3. Customising colours, labels, and layout

> **Note:** pyvis output relies on JavaScript (vis.js) which cannot execute
> inside VS Code's notebook viewer. Each cell writes an HTML file and opens it
> in your default browser. In Jupyter Lab/Notebook this renders inline.

````python
from venncy.renderers.pyvis import plot_venn2
from pathlib import Path
import webbrowser

def show_venn(path):
    """Open the generated HTML file in the default browser."""
    full_path = Path(path).resolve()
    webbrowser.open(f"file://{full_path}")

# Just provide the three areas — the renderer does the rest
net = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"), output_path="venn_basic.html")
show_venn("venn_basic.html")
````

### No overlap

````python
net = plot_venn2(area_a=10, area_b=3, area_ab=0, output_path="venn_no_overlap.html")
show_venn("venn_no_overlap.html")
````

### Full containment

````python
net = plot_venn2(area_a=10, area_b=3, area_ab=3, output_path="venn_containment.html")
show_venn("venn_containment.html")
````

### Customisation

You can pass custom colours, canvas dimensions, output path, and any extra `**kwargs` forwarded to vis.js node attributes. Use `text_kwargs` to customise the node font.

````python
net = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("rgba(50,200,50,0.35)", "rgba(200,50,50,0.35)"),
    canvas_width=1000,
    canvas_height=600,
    output_path="venn_custom.html",
    borderWidth=4,
    text_kwargs={"size": 24},
)
show_venn("venn_custom.html")
````

## 3-Circle Venn Diagram

### Basic usage

For three circles, provide six area values — the three individual areas (`area_a`, `area_b`, `area_c`) and the three pairwise overlaps (`area_ab`, `area_ac`, `area_bc`).

````python
from venncy.renderers.pyvis import plot_venn3

net = plot_venn3(
    area_a=10, area_b=8, area_c=6,
    area_ab=4, area_ac=3, area_bc=2,
    output_path="venn3_basic.html",
)
show_venn("venn3_basic.html")
````

### Customisation

Same options as `plot_venn2` — pass custom `colors`, `canvas_width`, `canvas_height`, or any `**kwargs` for vis.js nodes.

````python
net = plot_venn3(
    area_a=10, area_b=8, area_c=6,
    area_ab=4, area_ac=3, area_bc=2,
    labels=("A", "B", "C"),
    colors=("rgba(255,165,0,0.35)", "rgba(128,0,128,0.35)", "rgba(0,255,255,0.35)"),
    canvas_width=1000,
    canvas_height=600,
    output_path="venn3_custom.html",
    borderWidth=4,
    text_kwargs={"size": 24, "align": "center", "va": "middle"},
)
show_venn("venn3_custom.html")
````
