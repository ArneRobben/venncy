# pyvis renderer

Interactive vis.js–based Venn diagram that renders to a standalone HTML file.

```bash
pip install venncy[pyvis]
```

````python
from venncy.renderers.pyvis import plot_venn2

# Just provide the three areas — the renderer does the rest
net = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"), output_path="venn_basic.html")
print("Written to venn_basic.html")
````

## No overlap

````python
net = plot_venn2(area_a=10, area_b=3, area_ab=0, output_path="venn_no_overlap.html")
print("Written to venn_no_overlap.html")
````

## Full containment

````python
net = plot_venn2(area_a=10, area_b=3, area_ab=3, output_path="venn_containment.html")
print("Written to venn_containment.html")
````

## Customisation

You can pass custom colours, canvas dimensions, and output path.

````python
net = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("rgba(50,200,50,0.35)", "rgba(200,50,50,0.35)"),
    canvas_width=1000,
    canvas_height=600,
    output_path="venn_custom.html",
)
print("Written to venn_custom.html")
````
