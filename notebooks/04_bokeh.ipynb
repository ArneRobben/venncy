# Bokeh renderer

Interactive Venn diagrams with **Bokeh**.

```bash
pip install venncy[bokeh]
```

````python
from venncy.renderers.bokeh import plot_venn2
from bokeh.io import output_notebook, show

output_notebook()

# Just provide the three areas — the renderer does the rest
fig = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
show(fig)
````

## No overlap

````python
fig = plot_venn2(area_a=10, area_b=3, area_ab=0)
show(fig)
````

## Full containment

````python
fig = plot_venn2(area_a=10, area_b=3, area_ab=3)
show(fig)
````

## Customisation

You can pass custom colours, opacity, and figure dimensions.

````python
fig = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("green", "red"),
    alpha=0.25,
    width=700,
    height=500,
)
show(fig)
````
