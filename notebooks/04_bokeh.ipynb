# Bokeh Renderer

This notebook demonstrates how to create Venn diagrams using the **Bokeh** renderer.

**What you'll learn:**
1. Drawing 2-circle Venn diagrams (`plot_venn2`)
2. Drawing 3-circle Venn diagrams (`plot_venn3`)
3. Customising colours, labels, and layout

```bash
pip install venncy[bokeh]
```

````python
from venncy.renderers.bokeh import plot_venn2
from bokeh.io import output_notebook, show

output_notebook()

# Just provide the three areas — the renderer does the rest
fig = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
show(fig)
````

### No overlap

````python
fig = plot_venn2(area_a=10, area_b=3, area_ab=0)
show(fig)
````

### Full containment

````python
fig = plot_venn2(area_a=10, area_b=3, area_ab=3)
show(fig)
````

### Customisation

You can pass custom colours, opacity, figure dimensions, and any extra `**kwargs` forwarded to Bokeh's `ellipse` glyph. Use `text_kwargs` to customise the label.

````python
fig = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("green", "red"),
    alpha=0.25,
    width=700,
    height=500,
    line_color="black",
    line_width=3,
    text_kwargs={"text_font_size": "18pt"},
)
show(fig)
````

## 3-Circle Venn Diagram

### Basic usage

For three circles, provide six area values — the three individual areas (`area_a`, `area_b`, `area_c`) and the three pairwise overlaps (`area_ab`, `area_ac`, `area_bc`).

````python
from venncy.renderers.bokeh import plot_venn3

fig = plot_venn3(area_a=10, area_b=8, area_c=6, area_ab=4, area_ac=3, area_bc=2)
show(fig)
````

### Customisation

Same options as `plot_venn2` — pass custom `colors`, `alpha`, or any `**kwargs` for Bokeh's `ellipse` glyph.

````python
fig = plot_venn3(
    area_a=10, area_b=8, area_c=6,
    area_ab=4, area_ac=3, area_bc=2,
    labels=("A", "B", "C"),
    colors=("orange", "purple", "cyan"),
    alpha=0.4,
    line_color="black",
    line_width=3,
    text_kwargs={"text_font_size": "18pt"},
)
show(fig)
````
