# drawsvg Renderer

This notebook demonstrates how to create Venn diagrams using the **drawsvg** renderer.

**What you'll learn:**
1. Drawing 2-circle Venn diagrams (`plot_venn2`)
2. Drawing 3-circle Venn diagrams (`plot_venn3`)
3. Customising colours, labels, and layout

```bash
pip install venncy[drawsvg]
```

````python
from venncy.renderers.drawsvg import plot_venn2

# Just provide the three areas — the renderer does the rest
d = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
d
````

### No overlap

````python
d = plot_venn2(area_a=10, area_b=3, area_ab=0)
d
````

### Full containment

````python
d = plot_venn2(area_a=10, area_b=3, area_ab=3)
d
````

### Customisation

You can pass custom colours, opacity, dimensions, and any extra `**kwargs` forwarded to `drawsvg.Circle`. Use `text_kwargs` to customise the label.

````python
d = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("green", "red"),
    alpha=0.25,
    width=700,
    height=500,
    stroke="black",
    stroke_dasharray="5,5",
    text_kwargs={"fill": "darkblue"},
)
d
````

## 3-Circle Venn Diagram

### Basic usage

For three circles, provide six area values — the three individual areas (`area_a`, `area_b`, `area_c`) and the three pairwise overlaps (`area_ab`, `area_ac`, `area_bc`).

````python
from venncy.renderers.drawsvg import plot_venn3

d = plot_venn3(area_a=10, area_b=8, area_c=6, area_ab=4, area_ac=3, area_bc=2)
d
````

### Customisation

Same options as `plot_venn2` — pass custom `colors`, `alpha`, or any `**kwargs` for `drawsvg.Circle`.

````python
d = plot_venn3(
    area_a=10, area_b=8, area_c=6,
    area_ab=4, area_ac=3, area_bc=2,
    labels=("A", "B", "C"),
    colors=("orange", "purple", "cyan"),
    alpha=0.4,
    stroke="black",
    stroke_dasharray=".2, .25",
    text_kwargs={"fill": "darkblue"},
)
d
````

## Save to file

Export the SVG to a file — the output is resolution-independent.

````python
d = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("A", "B"))
d.save_svg("venn_example.svg")
print("Saved venn_example.svg")
````
