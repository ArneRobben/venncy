# drawsvg renderer

Lightweight SVG Venn diagrams with **drawsvg** — no heavy dependencies.

```bash
pip install venncy[drawsvg]
```

````python
from venncy.renderers.drawsvg import plot_venn2

# Just provide the three areas — the renderer does the rest
d = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("Dogs", "Cats"))
d
````

## No overlap

````python
d = plot_venn2(area_a=10, area_b=3, area_ab=0)
d
````

## Full containment

````python
d = plot_venn2(area_a=10, area_b=3, area_ab=3)
d
````

## Customisation

You can pass custom colours, opacity, and dimensions.

````python
d = plot_venn2(
    area_a=20,
    area_b=8,
    area_ab=5,
    labels=("Python", "R"),
    colors=("green", "red"),
    alpha=0.25,
    width=700,
    height=500,
)
d
````

## Save to file

Export the SVG to a file — the output is resolution-independent.

````python
d = plot_venn2(area_a=10, area_b=5, area_ab=2, labels=("A", "B"))
d.save_svg("venn_example.svg")
print("Saved venn_example.svg")
````
